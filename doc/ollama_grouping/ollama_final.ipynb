```python
import pandas as pd
import requests
```

## Batch 1 - Grouping & Deletion

```python
# Loading pre-processed CSV
df = pd.read_csv('final_pathologies_list.csv', header=None, skiprows=1)

# Extract relevant terms as plain text
terms = df.iloc[:, 0].dropna().astype(str).tolist()
term_block = "\n".join(terms[:112])
```

### gpt-oss:20b

```python
# Construct the prompt
system_prompt = (
    "You are a board-certified radiologist specializing in chest x-ray (CXR) findings."
    "Your task is to group clinically synonymous terms and identify unrelated ones."
)

user_prompt = f"""
Here is a list of terms from a chest x-ray dataset:

{term_block}

Instructions:
1. Group terms that are clinically equivalent or synonymous.
2. If a term is unrelated to chest pathology or indication, exclude it and list it separately.
3. Do not invent or add new terms.
4. Output format:

Final Group
1,"canonical_label":"<term>","members":["<term1>","<term2>",...]

Deleted Term
1,"<term>"
2,"<term>"
"""

# Send to Ollama
final_first_response_gpt = requests.post(
    'http://127.0.0.1:11434/api/chat',
    json={
        "model": "gpt-oss:20b",
        "messages": [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        "stream": False,
        "seed": 17,
        "temperature": 0.3,
        "top_p": 0.8,
        "num_predict": 512
    }
)
```

```python
print(final_first_response_gpt.json()["message"]["content"])
```

## Batch 2 - Grouping to Main 14 Class

### gpt-oss:20b

```python
# Construct the prompt
system_prompt = (
    "You are a board-certified radiologist specializing in chest x-ray (CXR) findings."
    "Your task is to group clinically synonymous terms and identify unrelated ones."
)

user_prompt = f"""
Here is a list of terms from a chest x-ray dataset:

{term_block}

Instructions:
1. Group those terms to these groups: Enlarged Cardiomediastinum, Cardiomegaly, Lung Opacity, Lung Lesion, Edema, Consolidation, Atelectasis, Pneumothorax, Pleural Effusion, Pleural Other, Fracture, Support Devices, and Pneumonia if they are clinically equivalent or synonymous.
2. If a term is unrelated to chest pathology or indication, exclude it and list it separately.
3. If a term is unrelated or not clinically equivalent to the groups, exclude it and list it separately.
4. Ensure that every term is assigned to exactly one of the following groups: groupings, unrelated terms, or deleted terms. No term should be left unclassified, and no term should appear in more than one category.
5. Do not invent or add new terms.
6. Output format:

Groupings
1,"Enlarged Cardiomediastinum","members":["<term1>","<term2>",...]

Unrelated Term (not clinically equivalent to the groups)
1,"<term>"
2,"<term>"

Deleted Term (not related to chest pathology or indication)
1,"<term>"
2,"<term>"
"""

# Send to Ollama
final_response_gpt = requests.post(
    'http://127.0.0.1:11434/api/chat',
    json={
        "model": "gpt-oss:20b",
        "messages": [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        "stream": False,
        "seed": 17,
        "temperature": 0.3,
        "top_p": 0.9,
        "num_predict": 1024
    }
)
```
