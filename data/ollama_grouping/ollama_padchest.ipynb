```python
import pandas as pd
import requests
```

# PadChest Dataset

## Batch 1

```python
# Loading pre-processed CSV
df = pd.read_csv('padchest_list.csv', header=None)

# Extract relevant terms as plain text
terms = df.iloc[:, 0].dropna().astype(str).tolist()
term_block = "\n".join(terms[:194])
```

### gpt-oss:20b

```python
# Construct the prompt
system_prompt = (
    "You are a board-certified radiologist specializing in chest x-ray (CXR) findings."
    "Your task is to group clinically synonymous terms and identify unrelated ones."
)

user_prompt = f"""
Here is a list of terms from a chest x-ray dataset:

{term_block}

Instructions:
1. Group terms that are clinically equivalent or synonymous.
2. If a term is unrelated to chest pathology or indication, exclude it and list it separately.
3. Do not invent or add new terms.
4. Output format:

Final Group
1,"canonical_label":"<term>","members":["<term1>","<term2>",...]

Deleted Term
1,"<term>"
2,"<term>"
"""

# Send to Ollama
padchest_first_response_gpt = requests.post(
    'http://127.0.0.1:11434/api/chat',
    json={
        "model": "gpt-oss:20b",
        "messages": [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        "stream": False,
        "seed": 17,
        "temperature": 0.3,
        "top_p": 0.8,
        "num_predict": 512
    }
)
```

```python
print(padchest_first_response_gpt.json()["message"]["content"])
```
